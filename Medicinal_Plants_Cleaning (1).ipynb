```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('01_suitability_scores (1).csv')
df.head()
```

```python
# 1. إظهار أبعاد الجدول (عدد الصفوف، عدد الأعمدة)
print("جدول البيانات يتكون من:")
print(f"عدد الصفوف: {df.shape[0]}")
print(f"عدد الأعمدة: {df.shape[1]}\n")

# 2. عرض ملخص الشامل للأعمدة وأنواع البيانات
df.info()
```

```python
# 1. حساب عدد القيم الفارغة في كل عمود
print("--- عدد القيم المفقودة في كل عمود ---")
print(df.isnull().sum())

# 2. حساب عدد الصفوف المكررة
print("\n--- عدد الصفوف المكررة بالكامل ---")
print(f"الصفووف المكررة: {df.duplicated().sum()}")
```

```python
# إحصائيات الأعمدة الرقمية (الأرقام)
print("--- إحصائيات الأعمدة الرقمية ---")
display(df.describe())

# إحصائيات الأعمدة النصية
print("--- إحصائيات الأعمدة النصية ---")
display(df.describe(include=['object']))
```

```python
# إظهار عدد الصفوف والأعمدة
print("أبعاد البيانات (صفوف، أعمدة):", df.shape)

# عرض تفاصيل الأعمدة وأنواع البيانات
df.info()
```

```python
# حساب القيم المفقودة في كل عمود
print("--- القيم المفقودة (Nulls) ---")
print(df.isnull().sum())

# حساب الصفوف المكررة
print("\nعدد الصفوف المكررة بالكامل:", df.duplicated().sum())
```

```python
# إحصائيات للأعمدة الرقمية والنصية معاً
df.describe(include='all')
```

```python
# 1. تعبئة القيم المفقودة في احتِياج الري بمتوسط العمود
df['irrigation_need_mm_year'] = df['irrigation_need_mm_year'].fillna(df['irrigation_need_mm_year'].mean())

# 2. تعبئة الملاحظات المفقودة بنص واضح
df['analyst_note'] = df['analyst_note'].fillna('No Note')

# التأكد من معالجة جميع القيم المفقودة
print("عدد القيم المفقودة بعد التنظيف:")
print(df[['irrigation_need_mm_year', 'analyst_note']].isnull().sum())
```

```python
# إظهار أنواع البيانات بعد التنظيف
df.dtypes
```

```python
# حفظ الملف المنظف بنفس المجلد
df.to_csv('01_suitability_scores_cleaned.csv', index=False)
print("تم حفظ الملف المنظف بنجاح باسم: 01_suitability_scores_cleaned.csv")
```

```python
# حساب متوسط درجة الملاءمة لكل نبات
top_plants = df.groupby('common_ar')['suitability_score'].mean().sort_values(ascending=False).head(10)

print("--- أعلى 10 نباتات ملاءمةً ---")
print(top_plants)
```

```python
# حساب متوسط الملاءمة لكل محافظة
top_govs = df.groupby('gov_ar')['suitability_score'].mean().sort_values(ascending=False)

print("--- ترتيب المحافظات حسب متوسط الملاءمة ---")
print(top_govs)
```

```python
# أكثر النباتات احتياجاً للمياه
water_needy = df.groupby('common_ar')['irrigation_need_mm_year'].mean().sort_values(ascending=False).head(5)

print("--- أكثر 5 نباتات احتياجاً للري (مم/سنة) ---")
print(water_needy)
```
