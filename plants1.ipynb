```python
def clean_life_span(val):
    terms = [t.strip().lower() for t in str(val).split(',')]

    valid_terms = []

    for term in terms:
        if 'annual' in term and 'annual' not in valid_terms:
            valid_terms.append('annual')
        elif 'biennial' in term and 'biennial' not in valid_terms:
            valid_terms.append('biennial')
        elif 'perennial' in term and 'perennial' not in valid_terms:
            valid_terms.append('perennial')

    return ', '.join(valid_terms) if valid_terms else 'Not Specified'


df['life_span'] = df['life_span'].apply(clean_life_span)

print(df[['scientific_name', 'life_span']].head(10))
```

```python
def clean_soil_texture(val):
    if pd.isna(val):
        return 'Not Specified'

    val_str = str(val).lower().strip()

    textures = []

    if 'light' in val_str or 'sand' in val_str:
        textures.append('light')

    if 'medium' in val_str or 'loam' in val_str or 'silt' in val_str:
        textures.append('medium')

    if 'heavy' in val_str or 'clay' in val_str:
        textures.append('heavy')

    if 'organic' in val_str or 'humus' in val_str:
        textures.append('organic')

    return ', '.join(textures) if textures else 'Not Specified'


df['soil_texture'] = df['soil_texture'].apply(clean_soil_texture)

print(df[['scientific_name', 'soil_texture']].head(10))
```

```python
def clean_salinity(val):
    if pd.isna(val):
        return 'Not Specified'

    val_str = str(val).lower().strip()

    less_than_match = re.search(
        r'<\s*(\d+\.?\d*)\s*ds/m',
        val_str
    )

    if less_than_match:
        threshold = float(less_than_match.group(1))

        if threshold <= 4:
            return 'Low (<4 dS/m)'
        elif threshold <= 10:
            return 'Moderate (4-10 dS/m)'
        else:
            return 'High / Tolerant (>10 dS/m)'

    if 'low' in val_str or 'glycophyte' in val_str:
        return 'Low (<4 dS/m)'

    elif 'moderate' in val_str or '6 ds/m' in val_str:
        return 'Moderate (4-10 dS/m)'

    elif (
        'tolerant' in val_str
        or 'ec 0.128-15.3' in val_str
        or 'sodicity' in val_str
    ):
        return 'High / Tolerant (>10 dS/m)'

    else:
        return 'Not Specified'


df['soil_salinity_clean'] = df['soil_salinity'].apply(
    clean_salinity
)

print(
    df[
        [
            'scientific_name',
            'soil_salinity',
            'soil_salinity_clean'
        ]
    ].head(10)
)
```

```python
def classify_waterlogging_sensitivity(val):
    val_str = str(val).lower()

    if (
        'fatal' in val_str
        or 'not waterlogged' in val_str
        or 'no waterlogging' in val_str
    ):
        return 'High (waterlogging intolerant)'

    return 'Moderate (standard well-drained need)'
```

```python
def classify_drought_tolerance(val):
    val_str = str(val).lower()

    if (
        'moisture retentive' in val_str
        or 'moist but' in val_str
    ):
        return 'Low (needs consistent moisture)'

    if (
        'dry spells' in val_str
        or 'excessive (dry' in val_str
        or 'free to excessive' in val_str
    ):
        return 'High (tolerates dry conditions)'

    return 'Not specified'
```

```python
df['waterlogging_sensitivity'] = (
    df['soil_drainage']
    .apply(classify_waterlogging_sensitivity)
)

df['drought_tolerance'] = (
    df['soil_drainage']
    .apply(classify_drought_tolerance)
)

print(
    df[
        [
            'scientific_name',
            'soil_drainage',
            'waterlogging_sensitivity',
            'drought_tolerance'
        ]
    ].to_string(index=False)
)
```

```python
def clean_light_intensity(val):
    if pd.isna(val) or str(val).strip() == '':
        return 'Not Specified'
    
    val_lower = str(val).lower().strip()

    if 'moderate shade' in val_lower:
        return 'Moderate Shade'
    
    elif (
        'partial shade' in val_lower
        or 'light shade' in val_lower
    ) and not (
        'full sun' in val_lower
        or 'very bright' in val_lower
    ):
        return 'Partial Shade'
    
    elif (
        'very bright' in val_lower
        or 'clear skies' in val_lower
        or 'full sun' in val_lower
    ):
        return 'Full Sun / Bright'
    
    elif 'cloudy' in val_lower:
        return 'Full Sun / Bright (tolerates cloudy days)'
    
    return 'Not Specified'


df['light_intensity_clean'] = (
    df['light_intensity']
    .apply(clean_light_intensity)
    .astype('category')
)

print("Cleaned light_intensity Distribution:")
print(df['light_intensity_clean'].value_counts(dropna=False))
```

```python


# 2. Define function to clean climate_zone column
def clean_climate_zone(val):
    if pd.isna(val) or str(val).strip() == '':
        return 'Not Specified'
    
    # Split by '|' to strip out attached Photoperiod text and take the climate part
    climate_part = str(val).split('|')[0].strip()
    
    return climate_part if climate_part else 'Not Specified'

# 3. Create the new cleaned column while keeping the original
df['climate_zone_cleaned'] = df['climate_zone'].apply(clean_climate_zone)

# 4. Display sample of original vs cleaned data
print(df[['climate_zone', 'climate_zone_cleaned']].head())

```

```python
def clean_photoperiod(val):
    if pd.isna(val) or str(val).strip() == '':
        return 'Not Specified'
    
    val_str = str(val).lower()
    
    # Check present terms
    has_short = 'short day' in val_str
    has_neutral = 'neutral day' in val_str
    has_long = 'long day' in val_str
    
    # Classify photoperiod types
    if has_short and has_neutral and has_long:
        return 'Flexible / Any'
    elif has_short and not has_long:
        return 'Short Day (<12h)'
    elif has_long and not has_short:
        return 'Long Day (>14h)'
    elif has_neutral:
        return 'Day Neutral (12-14h)'
    
    return 'Not Specified'

# 3. Create the new cleaned column alongside the original one
df['photoperiod_cleaned'] = df['photoperiod'].apply(clean_photoperiod)

# 4. Define category order for analysis
photoperiod_order = ['Short Day (<12h)', 'Day Neutral (12-14h)', 'Long Day (>14h)', 'Flexible / Any', 'Not Specified']
df['photoperiod_cleaned'] = pd.Categorical(df['photoperiod_cleaned'], categories=photoperiod_order, ordered=True)

# 5. Verify the results
print("Original vs Cleaned Comparison:")
print(df[['photoperiod', 'photoperiod_cleaned']].drop_duplicates())

print("\nDistribution of Cleaned Photoperiods:")
print(df['photoperiod_cleaned'].value_counts(dropna=False))

```

```python
def clean_killing_temp_rest(val):
    if pd.isna(val) or str(val).strip() == '':
        return 'Not Specified'
    
    # Search for the first integer/float number (including negative numbers like -5, -10)
    match = re.search(r'(-?\d+(\.\d+)?)', str(val))
    if match:
        return match.group(1)
    
    return 'Not Specified'

# 3. Create the new cleaned column while preserving the original
df['killing_temp_rest_cleaned'] = df['killing_temp_rest'].apply(clean_killing_temp_rest)

# 4. Display sample comparison of original vs cleaned data
print(df[['killing_temp_rest', 'killing_temp_rest_cleaned']].drop_duplicates())
```

```python
def clean_killing_temp_growth(val):
    if pd.isna(val) or str(val).strip() == '':
        return 'Not Specified'
    
    # Search for numeric values (e.g., -2, 0, or positive floats)
    match = re.search(r'(-?\d+(\.\d+)?)', str(val).strip())
    if match:
        return match.group(1)
    
    return 'Not Specified'

# 3. Create the new cleaned column while keeping the original intact
df['killing_temp_growth_cleaned'] = df['killing_temp_growth'].apply(clean_killing_temp_growth)

# 4. Verify the comparison results
print("Original vs Cleaned Comparison:")
print(df[['killing_temp_growth', 'killing_temp_growth_cleaned']].drop_duplicates())
```

```python
def extract_crop_cycle_min(row):
    val = row['production_system']
    if pd.notna(val):
        match = re.search(r'Crop cycle\s*\|\s*(\d+)\s*\|\s*(\d+)', str(val), re.IGNORECASE)
        if match and (int(match.group(1)) > 0 or int(match.group(2)) > 0):
            return int(match.group(1))
            
    cycle_val = row['crop_cycle_days']
    if pd.notna(cycle_val):
        cycle_str = str(cycle_val).strip()
        if '|' in cycle_str:
            parts = [p.strip() for p in cycle_str.split('|')]
            if len(parts) >= 1 and parts[0].isdigit():
                return int(parts[0])
        elif cycle_str.isdigit():
            return int(cycle_str)
    return None

# 2. Improved function to handle max days with zero-check
def extract_crop_cycle_max(row):
    val = row['production_system']
    if pd.notna(val):
        match = re.search(r'Crop cycle\s*\|\s*(\d+)\s*\|\s*(\d+)', str(val), re.IGNORECASE)
        if match and (int(match.group(1)) > 0 or int(match.group(2)) > 0):
            return int(match.group(2))
            
    cycle_val = row['crop_cycle_days']
    if pd.notna(cycle_val):
        cycle_str = str(cycle_val).strip()
        if '|' in cycle_str:
            parts = [p.strip() for p in cycle_str.split('|')]
            if len(parts) >= 2 and parts[1].isdigit():
                return int(parts[1])
        elif cycle_str.isdigit():
            return int(cycle_str)
    return None

# 3. Apply functions
df['crop_cycle_min_days'] = df.apply(extract_crop_cycle_min, axis=1)
df['crop_cycle_max_days'] = df.apply(extract_crop_cycle_max, axis=1)

# 4. Display all rows fully without truncation
pd.set_option('display.max_rows', None)      # Display all rows
pd.set_option('display.max_columns', None)   # Display all columns
pd.set_option('display.width', 1000)         # Prevent line breaks

check_cols = ['scientific_name', 'production_system', 'crop_cycle_days', 'crop_cycle_min_days', 'crop_cycle_max_days']
print(df[check_cols])
```
