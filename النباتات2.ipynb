```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\A-FF-L1-D07\Desktop\03_Plants_النباتات-20260830T162732Z-1-001\03_Plants_النباتات\02_target_plants_list.csv")
```

```python
df['scientific_name'] = df['scientific_name'].str.strip()
```

```python
df['common_en'] = df['common_en'].str.strip()
```

```python

df['common_ar'] = df['common_ar'].str.strip()
```

```python
df['common_name'] = df['common_ar'].fillna('') + ' (' + df['common_en'].fillna('') + ')'

df['common_name'] = df['common_name'].str.strip()

print(df[['common_ar', 'common_en', 'common_name']].head())
```

```python
df['family'] = df['family'].str.strip()
```

```python
df['key_api'] = df['key_api'].str.strip()
```

```python
df['api_use'] = df['api_use'].str.strip()
```

```python
df['hs_codes'] = df['hs_codes'].str.strip()
```

```python
df[['hs_code_raw', 'hs_code_extract']] = df['hs_codes'].str.split(';', expand=True)

# Clean any leading or trailing spaces in the new columns
df['hs_code_raw'] = df['hs_code_raw'].str.strip()
df['hs_code_extract'] = df['hs_code_extract'].str.strip()

# Preview the results
print(df[['plant_id', 'hs_codes', 'hs_code_raw', 'hs_code_extract']].head())
```

```python
df['notes'] = df['notes'].fillna('Not specified').str.strip()
```

```python
df.to_csv("02_target_plants_list_cleaned.csv", index=False, encoding='utf-8-sig')
```
