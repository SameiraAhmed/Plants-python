```python
print("Missing Values:")
display(df_geo.isnull().sum())

print("\nDuplicate Rows:")
print(df_geo.duplicated().sum())
```

```python
df_geo.columns = (
    df_geo.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print(df_geo.columns)
```

```python
text_columns = [
    "adm1_pcode",
    "gov_en",
    "gov_ar"
]

for col in text_columns:
    df_geo[col] = (
        df_geo[col]
        .astype(str)
        .str.strip()
    )
```

```python
print(df_geo["gov_ar"].unique())
```

```python
print(df_geo.dtypes)
```

```python
numeric_columns = [
    "lat",
    "lon",
    "area_sqkm"
]

for col in numeric_columns:
    df_geo[col] = pd.to_numeric(
        df_geo[col],
        errors="coerce"
    )
```

```python
display(df_geo.isnull().sum())
```

```python
invalid_lat = df_geo[
    (df_geo["lat"] < -90) |
    (df_geo["lat"] > 90)
]

invalid_lon = df_geo[
    (df_geo["lon"] < -180) |
    (df_geo["lon"] > 180)
]

print("Invalid Latitude:", len(invalid_lat))
print("Invalid Longitude:", len(invalid_lon))
```

```python
invalid_area = df_geo[
    df_geo["area_sqkm"] <= 0
]

print("Invalid Area:", len(invalid_area))
```

```python
df_geo["lat"] = df_geo["lat"].round(6)

df_geo["lon"] = df_geo["lon"].round(6)

df_geo["area_sqkm"] = df_geo["area_sqkm"].round(2)
```

```python
df_geo.to_csv(
    "Geography_Cleaned.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Geography_Cleaned.csv saved successfully")
```
